```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt


black_img = np.zeros(shape=(512,512,3), dtype=np.int16)
plt.imshow(black_img)
```

```python
# perhatikan koordinat titik2 pt1 dan pt2
cv.rectangle(black_img, pt1=(384,0), pt2=(510,150), color=(0,255,0), thickness=10)
plt.imshow(black_img)
```

```python
cv.rectangle(black_img, pt1=(200,200), pt2=(300,300), color=(0,0,255), thickness=15)
plt.imshow(black_img)

cv.circle(black_img, center=(100,100), radius=50, color=(255,0,0), thickness=8)
plt.imshow(black_img)
```

```python
cv.line(black_img, pt1=(0,0), pt2=(512,512), color=(255,255,255), thickness=5)
plt.imshow(black_img)

font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(black_img, text='Hello', org=(10,500), fontFace=font, fontScale=4, color=(255,255,0), thickness=2, lineType=cv.LINE_AA)
plt.imshow(black_img)
```

```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt


black_img2 = np.zeros(shape=(512,512,3), dtype=np.int32)
plt.imshow(black_img2)


vertices = np.array([[100,300],[200,200],[400,300],[200,400]], dtype=np.int32)

pts = vertices.reshape((-1,1,2))
pts


cv.polylines(black_img2, [pts], isClosed=True, color=(255,0,0), thickness=5)
plt.imshow(black_img2)
```

```python
# Tampilkan Red-Blue (Nol-kan Green/index 1)
img_RB = imgRGB.copy()
img_RB[:, :, 1] = 0

# Tampilkan Green-Blue (Nol-kan Red/index 0)
img_GB = imgRGB.copy()
img_GB[:, :, 0] = 0

plt.subplot(1, 2, 1); plt.imshow(img_RB); plt.title("Red-Blue")
plt.subplot(1, 2, 2); plt.imshow(img_GB); plt.title("Green-Blue")
plt.show()
```

```python
img_crop = imgRGB[20:115, 25:120]
plt.imshow(img_crop)
plt.title("Cropping Baris & Kolom")
plt.show()
```

```python
img_red_only = imgRGB[5:30, :].copy()
img_red_only[:, :, 1] = 0 # Hilangkan Green
img_red_only[:, :, 2] = 0 # Hilangkan Blue
plt.imshow(img_red_only)
plt.title("Baris 5-30, Channel Red Saja")
plt.show()
```

```python
import random

img_kotak = imgRGB.copy()
h, w = img_kotak.shape[:2]

for _ in range(5):
    # Kordinat acak
    x1, y1 = random.randint(0, w-50), random.randint(0, h-50)
    x2, y2 = x1 + random.randint(20, 100), y1 + random.randint(20, 100)
    # Warna acak RGB
    r, g, b = random.randint(0, 255), random.randint(0, 255), random.randint(0, 255)

    cv2.rectangle(img_kotak, (x1, y1), (x2, y2), (r, g, b), 3)

plt.imshow(img_kotak)
plt.title("5 Kotak Random")
plt.show()
```

```python
# Angka -1 buat muter vertikal dan horizontal sekaligus
img_flip = cv2.flip(imgRGB, -1)
plt.imshow(img_flip)
plt.title("Image Terbalik")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt


img = cv2.imread("/content/drive/MyDrive/PCVK/GAMBAR/kitten.jpg")
img_RGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_B = img_RGB.copy()
img_B[:, :, 0] = 0
img_B[:, :, 1] = 0


h, w = img_B.shape[:2]


cv2.rectangle(img_B, pt1=(0, h - 80), pt2=(250, h), color=(255, 255, 255), thickness=4)

font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(img_B, text="kitten.jpg", org=(15, h - 30), fontFace=font, fontScale=1.2, color=(255, 255, 255), thickness=2)

plt.imshow(img_B)
plt.title("Tugas 8 & 9: Channel B + Kotak & Teks")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

ktm_img = cv2.imread("/content/drive/MyDrive/PCVK/GAMBAR/ktm.jpeg")
ktm_rgb = cv2.cvtColor(ktm_img, cv2.COLOR_BGR2RGB)


ktm_crop = ktm_rgb[370:820, 30:650].copy()

ktm_sensor = ktm_crop.copy()

cv2.rectangle(ktm_sensor, pt1=(140, 320), pt2=(520, 365), color=(255, 0, 0), thickness=-1)

cv2.rectangle(ktm_sensor, pt1=(140, 370), pt2=(370, 390), color=(0, 0, 255), thickness=-1)

cv2.rectangle(ktm_sensor, pt1=(140, 395), pt2=(460, 415), color=(0, 255, 0), thickness=-1)

plt.figure(figsize=(8, 5))
plt.imshow(ktm_sensor)
plt.title("Tugas Kelompok 1: Sensor NIM, Nama, Prodi")
plt.show()
```
